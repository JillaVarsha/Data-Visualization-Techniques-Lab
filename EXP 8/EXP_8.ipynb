```python
import pandas as pd
import numpy as np

# Load dataset
arrivals = pd.read_csv(
    "Inbound Tourism-Arrivals.csv",
    encoding="latin1"
)

# Fill country names downward
arrivals["Country"] = arrivals["Basic data and indicators"].ffill()

# Combine the different measure columns
arrivals["Measure"] = (
    arrivals["Unnamed: 5"]
    .combine_first(arrivals["Unnamed: 6"])
    .combine_first(arrivals["Unnamed: 7"])
)

# Select year columns
year_columns = [str(year) for year in range(1995, 2023)]

# Keep only rows that contain an actual measure
arrivals_clean = arrivals[
    arrivals["Measure"].notna()
].copy()

# Remove unnecessary rows such as category/header rows
valid_measures = [
    "Total arrivals",
    "Overnights visitors (tourists)",
    "Same-day visitors (excursionists)",
    "of which, cruise passengers"
]

arrivals_clean = arrivals_clean[
    arrivals_clean["Measure"].isin(valid_measures)
].copy()

# Keep required columns
arrivals_clean = arrivals_clean[
    ["Country", "Measure", "Units"] + year_columns
]

# Convert wide format to long format
arrivals_clean = arrivals_clean.melt(
    id_vars=["Country", "Measure", "Units"],
    value_vars=year_columns,
    var_name="Year",
    value_name="Value"
)

# Replace ".." with missing values
arrivals_clean["Value"] = arrivals_clean["Value"].replace(
    ["..", "...", "—", "-"],
    np.nan
)

# Remove commas and convert values to numeric
arrivals_clean["Value"] = (
    arrivals_clean["Value"]
    .astype(str)
    .str.replace(",", "", regex=False)
)

arrivals_clean["Value"] = pd.to_numeric(
    arrivals_clean["Value"],
    errors="coerce"
)

# Convert Year to integer
arrivals_clean["Year"] = arrivals_clean["Year"].astype(int)

# Clean country names
arrivals_clean["Country"] = arrivals_clean["Country"].str.strip()

# Clean Units
arrivals_clean["Units"] = arrivals_clean["Units"].str.strip()

# Remove completely empty values
arrivals_clean = arrivals_clean.dropna(
    subset=["Value"]
).reset_index(drop=True)

# Display the cleaned dataset
print("Cleaned Dataset Shape:", arrivals_clean.shape)

display(arrivals_clean.head(20))

# Display information
print("\nDataset Information:")
arrivals_clean.info()

# Save cleaned dataset
arrivals_clean.to_csv(
    "Cleaned_Inbound_Tourism_Arrivals.csv",
    index=False
)

print("\nCleaned file saved successfully!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load cleaned arrivals dataset
arrivals = pd.read_csv(
    "Cleaned_Inbound_Tourism_Arrivals.csv"
)

# Keep only total arrivals
total_arrivals = arrivals[
    arrivals["Measure"] == "Total arrivals"
].copy()

# Calculate total arrivals worldwide for each year
global_arrivals = (
    total_arrivals
    .groupby("Year")["Value"]
    .sum()
    .reset_index()
)

print("GLOBAL TOURISM ARRIVALS BY YEAR")
display(global_arrivals)

# Plot global tourism arrivals trend
plt.figure(figsize=(12, 6))

plt.plot(
    global_arrivals["Year"],
    global_arrivals["Value"],
    marker="o"
)

plt.title("Global Tourism Arrivals (1995–2022)")
plt.xlabel("Year")
plt.ylabel("International Tourist Arrivals (Thousands)")
plt.grid(True)
plt.xticks(range(1995, 2023, 2))

plt.show()
```

```python
# Compare important tourism years

covid_years = global_arrivals[
    global_arrivals["Year"].isin([2019, 2020, 2021, 2022])
].copy()

display(covid_years)

# Get values
value_2019 = global_arrivals.loc[
    global_arrivals["Year"] == 2019, "Value"
].iloc[0]

value_2020 = global_arrivals.loc[
    global_arrivals["Year"] == 2020, "Value"
].iloc[0]

value_2021 = global_arrivals.loc[
    global_arrivals["Year"] == 2021, "Value"
].iloc[0]

value_2022 = global_arrivals.loc[
    global_arrivals["Year"] == 2022, "Value"
].iloc[0]

# Calculate percentage changes
decline_2020 = ((value_2020 - value_2019) / value_2019) * 100
change_2021 = ((value_2021 - value_2020) / value_2020) * 100
change_2022 = ((value_2022 - value_2021) / value_2021) * 100
recovery_2022 = ((value_2022 - value_2019) / value_2019) * 100

print(f"2019 → 2020 change: {decline_2020:.2f}%")
print(f"2020 → 2021 change: {change_2021:.2f}%")
print(f"2021 → 2022 change: {change_2022:.2f}%")
print(f"2022 compared with 2019: {recovery_2022:.2f}%")
```

```python
# Top tourism destinations in 2019
top_2019 = (
    total_arrivals[
        total_arrivals["Year"] == 2019
    ]
    .sort_values("Value", ascending=False)
    .head(15)
)

print("TOP 15 TOURISM DESTINATIONS - 2019")

display(
    top_2019[
        ["Country", "Value", "Units"]
    ]
)
```

```python
plt.figure(figsize=(12, 7))

plt.barh(
    top_2019["Country"],
    top_2019["Value"]
)

plt.title("Top 15 Tourism Destinations by Arrivals - 2019")
plt.xlabel("International Tourist Arrivals (Thousands)")
plt.ylabel("Country")

plt.gca().invert_yaxis()

plt.show()
```

```python
# Create 2019 and 2022 arrival tables
arrivals_2019 = total_arrivals[
    total_arrivals["Year"] == 2019
][["Country", "Value"]].rename(
    columns={"Value": "Arrivals_2019"}
)

arrivals_2022 = total_arrivals[
    total_arrivals["Year"] == 2022
][["Country", "Value"]].rename(
    columns={"Value": "Arrivals_2022"}
)

# Merge the two years
recovery = pd.merge(
    arrivals_2019,
    arrivals_2022,
    on="Country",
    how="inner"
)

# Calculate recovery percentage
recovery["Recovery_Percentage"] = (
    recovery["Arrivals_2022"] /
    recovery["Arrivals_2019"]
) * 100

# Calculate percentage change
recovery["Change_Percentage"] = (
    (recovery["Arrivals_2022"] -
     recovery["Arrivals_2019"])
    / recovery["Arrivals_2019"]
) * 100

# Remove invalid values
recovery = recovery[
    recovery["Arrivals_2019"] > 0
].copy()

# Countries with highest recovery
top_recovery = recovery.sort_values(
    "Recovery_Percentage",
    ascending=False
).head(15)

print("TOP 15 COUNTRIES BY TOURISM RECOVERY")
display(
    top_recovery[
        [
            "Country",
            "Arrivals_2019",
            "Arrivals_2022",
            "Recovery_Percentage",
            "Change_Percentage"
        ]
    ]
)
```

```python
plt.figure(figsize=(12, 7))

plt.barh(
    top_recovery["Country"],
    top_recovery["Recovery_Percentage"]
)

plt.axvline(
    100,
    linestyle="--"
)

plt.title("Top 15 Countries by Tourism Recovery: 2022 vs 2019")
plt.xlabel("2022 Arrivals as % of 2019 Arrivals")
plt.ylabel("Country")

plt.gca().invert_yaxis()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load cleaned expenditure dataset
expenditure = pd.read_csv(
    "Cleaned_Inbound_Tourism_Expenditure.csv"
)

# Display available measures
print("Available Expenditure Measures:")
print(expenditure["Measure"].unique())

# Calculate global expenditure by measure and year
global_expenditure = (
    expenditure
    .groupby(["Year", "Measure"])["Value"]
    .sum()
    .reset_index()
)

print("\nGLOBAL TOURISM EXPENDITURE")
display(global_expenditure)

# Plot expenditure by measure
plt.figure(figsize=(12, 6))

for measure in global_expenditure["Measure"].unique():

    data = global_expenditure[
        global_expenditure["Measure"] == measure
    ]

    plt.plot(
        data["Year"],
        data["Value"],
        marker="o",
        label=measure
    )

plt.title("Global Tourism Expenditure by Category (1995–2022)")
plt.xlabel("Year")
plt.ylabel("Expenditure (US$ Millions)")
plt.legend()
plt.grid(True)
plt.xticks(range(1995, 2023, 2))

plt.show()
```

```python
# COVID-era expenditure comparison

expenditure_covid = global_expenditure[
    global_expenditure["Year"].isin([2019, 2020, 2021, 2022])
].copy()

display(expenditure_covid)

# Calculate percentage change for each expenditure category
for measure in expenditure_covid["Measure"].unique():

    data = expenditure_covid[
        expenditure_covid["Measure"] == measure
    ].set_index("Year")

    value_2019 = data.loc[2019, "Value"]
    value_2020 = data.loc[2020, "Value"]
    value_2022 = data.loc[2022, "Value"]

    decline = (
        (value_2020 - value_2019)
        / value_2019
    ) * 100

    recovery = (
        (value_2022 - value_2019)
        / value_2019
    ) * 100

    print("\n", measure)
    print(f"2019 → 2020: {decline:.2f}%")
    print(f"2022 compared with 2019: {recovery:.2f}%")
```
