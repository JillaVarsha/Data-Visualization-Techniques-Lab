```python
import pandas as pd
import numpy as np

df = pd.read_csv("olympics_singles_agg.csv")

print("Original Shape:", df.shape)

df["final_rank"] = pd.to_numeric(df["final_rank"], errors="coerce")
df["fs_rank"] = pd.to_numeric(df["fs_rank"], errors="coerce")

df["tech_dominance"] = (
    df["tech_dominance"]
    .str.replace("%", "", regex=False)
)

df["tech_dominance"] = pd.to_numeric(
    df["tech_dominance"], errors="coerce"
)

df["gender"] = df["gender"].str.strip()
df["skater"] = df["skater"].str.strip()
df["nation"] = df["nation"].str.strip()

df.drop_duplicates(inplace=True)

print("Missing Values:")
print(df.isnull().sum())

print("Cleaned Shape:", df.shape)

df.to_csv("Cleaned_Olympic_Figure_Skating.csv", index=False)

print("\nCleaned dataset saved successfully!")
print(df.head())
```
