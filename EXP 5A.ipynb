```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_excel("Retail_Sales_Raw .xlsx")

df["OrderDate"] = pd.to_datetime(df["OrderDate"], dayfirst=False, errors="coerce")
df["Sales"] = df["Quantity"] * df["UnitPrice"] * (1 - df["Discount"].fillna(0))

monthly_sales = df.groupby(df["OrderDate"].dt.strftime("%B"))["Sales"].sum()

months = monthly_sales.index
sales = monthly_sales.values
```

```python
plt.plot(months, sales, marker="o")
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.bar(months, sales)
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.pie(sales, labels=months, autopct="%1.1f%%")
plt.title("Sales Distribution")
plt.show()
```

```python
plt.hist(df["Sales"].dropna(), bins=5)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.scatter(df["Quantity"], df["Sales"])
plt.title("Quantity vs Sales")
plt.xlabel("Quantity")
plt.ylabel("Sales")
plt.show()
```

```python
plt.boxplot(df["Sales"].dropna())
plt.title("Sales Box Plot")
plt.ylabel("Sales")
plt.show()
```

```python
monthly_quantity = df.groupby(df["OrderDate"].dt.strftime("%B"))["Quantity"].sum()

plt.plot(months, sales, label="Sales")
plt.plot(months, monthly_quantity.values, label="Quantity")
plt.title("Sales and Quantity")
plt.xlabel("Months")
plt.ylabel("Value")
plt.xticks(rotation=45)
plt.legend()
plt.show()
```
